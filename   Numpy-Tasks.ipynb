## TWO: Weather Data

Create the following array:
 temperatures = np.array([
   18, 21, 25, 17, 16,
   23, 28, 30, 22, 19,
   15, 24, 27, 20
])

Find:

Days above 25 degrees
Days below 20 degrees
Avg temp
Hottest day
Coldest day


How many days exceeded the avg temp?

```python
temperatures = np.array([    18, 21, 25, 17, 16,    23, 28, 30, 22, 19,    15, 24, 27, 20 ])
np.sum(temperatures > 25)
```

```python
np.sum(temperatures < 20)
```

```python
temperatures.mean()
```
